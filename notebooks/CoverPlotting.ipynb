```python
script_dir = os.path.dirname(os.path.abspath(__file__)) if "__file__" in globals() else os.getcwd()

# Path to the CSV file
csv_path = os.path.join(script_dir, "../data/architectural_digest_covers.csv")

# Path to the SQLite database
db_path = os.path.join(script_dir, "../data/architectural_digest_covers.db")

# Read CSV file
df = pd.read_csv(csv_path)

# Connect to SQLite database
conn = sqlite3.connect(db_path)

# Clear table if it exists
conn.execute("DROP TABLE IF EXISTS covers")

# Write dataframe to SQLite table
df.to_sql('covers', conn, if_exists='replace', index=False)
```

```python
springQuery = "SELECT primary_color FROM covers WHERE season = 'Spring' ORDER BY year ASC"
springResult = pd.read_sql_query(springQuery, conn)
SpringColors = springResult['primary_color'].tolist()

summerQuery = "SELECT primary_color FROM covers WHERE season = 'Summer' ORDER BY year ASC"
summerResult = pd.read_sql_query(summerQuery, conn)
SummerColors = summerResult['primary_color'].tolist()

fallQuery = "SELECT primary_color FROM covers WHERE season = 'Fall' ORDER BY year ASC"
fallResult = pd.read_sql_query(fallQuery, conn)
FallColors = fallResult['primary_color'].tolist()

winterQuery = "SELECT primary_color FROM covers WHERE season = 'Winter' ORDER BY year ASC"
winterResult = pd.read_sql_query(winterQuery, conn)
WinterColors = winterResult['primary_color'].tolist()
```

```python
waffleQuery = "SELECT cover_image_url FROM covers WHERE year='1929' AND month='January'"
waffleResult = pd.read_sql_query(waffleQuery, conn)
cover_image_url = waffleResult['cover_image_url'][0]

color_analysis = ColorAnalysis.ColorAnalysis(cover_image_url)
palette = color_analysis.get_palette_NEW()

primary_color_name = ColorConversion.ColorConversion.get_color_name(palette[0][0])
print("PRIMARY:" + primary_color_name)
for i, secondary_colors in enumerate(palette[1:], start=1):
    color_name = ColorConversion.ColorConversion.get_color_name(secondary_colors[0])
    print("SECONDARY " + str(i) + ":" + color_name)


print (cover_image_url)
for x in palette:
    print(x)



# palette is now a list of (rgb_array, proportion) tuples, e.g. (array([r,g,b]), 0.3039)
sorted_palette = sorted(palette, key=lambda item: item[1], reverse=True)

# Convert proportions to exactly 100 cells using largest-remainder allocation
raw_cells = [pct * 100 for _, pct in sorted_palette]
cell_counts = [int(np.floor(v)) for v in raw_cells]
remaining = 100 - sum(cell_counts)

if remaining > 0:
    remainders = [v - np.floor(v) for v in raw_cells]
    for idx in np.argsort(remainders)[::-1][:remaining]:
        cell_counts[idx] += 1

fig, ax = plt.subplots(figsize=(10, 10))
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_aspect('equal')
ax.invert_yaxis()
ax.axis('off')

cell_index = 0
for (color, _), cells_to_color in zip(sorted_palette, cell_counts):
    rgb_normalized = tuple(float(c) / 255.0 for c in color)

    for _ in range(cells_to_color):
        row = cell_index // 10
        col = cell_index % 10
        rect = plt.Rectangle((col, row), 1, 1, facecolor=rgb_normalized, edgecolor='white', linewidth=0.5)
        ax.add_patch(rect)

        cell_index += 1
        if cell_index >= 100:
            break

    if cell_index >= 100:
        break

plt.tight_layout()
# plt.savefig('../data/plots/color_grid.png', dpi=300, bbox_inches='tight')
plt.show()
```
